## Ejercicio 4

Implementa un programa que simule una jugada de **tirar dos dados**. El programa debe pedir al usuario dos valores enteros y almacenarlos en las variables ```d1``` y ```d2```.

En primer lugar, se debe verificar que ambos valores son **correctos**, es decir, están entre los posibles de un dado de seis caras (1 a 6). En caso contrario, el programa debe informar cuál de ellos es incorrecto (pueden ser uno de ambos, o los dos).

Si ambos valores introducidos son correctos, entonces se debe escribir por pantalla el **resultado de la jugada**, entre los siguientes, en función de la suma de ambos:
* 7 u 11, el usuario gana.
* 2, 3 o 12, el usuario pierde.
* 4, 5, 6, 8, 9 o 10, el usuario puede lanzar los dados de nuevo.



````python
try_again = True
while try_again:
  n1 = -1
  n1_valid = False
  while not n1_valid:
    n1 = int(input("Dice 1) Introduce un número entre 1 y 6: "))
    if(n1 <= 6 and n1 >= 1): n1_valid = True
  n2 = -1
  n2_valid = False
  while not n2_valid:
    n2 = int(input("Dice 2) Introduce un número entre 1 y 6: "))
    if(n2 <= 6 and n2 >= 1): n2_valid = True
  
  if((n1 + n2) in [4,5,6,8,9,10]):
    print("Throw the dice again, n1=", n1, " n2=", n2)
  elif ((n1 + n2) in [7, 11]):
    print("N1 WINS!"); try_again = False
  elif ((n1 + n2) in [2, 3, 12]):
    print("N1 WINS!"); try_again = False
````

> **NOTA:** Para saber si una cadena ```cad``` incluye solo dígtos o no, puede utilizarse el método ```cad.isdigit()```.

## Ejercicio 13

Implementa un programa que reciba como entrada un número entero positivo ```n``` y lo **factorice**, es decir, imprima por pantalla la secuencia creciente de sus factores o **divisores primos**. Por ejemplo, la secuencia de factores de 126 es: 2, 3, 3, 7.


````python
n1 = int(input("Introduce un número positivo mayor a 0: "))
result = []
for i in range(2, int(n1 ** 0.5) + 1, 1):
  while n1 % i == 0:
    result.append(i)
    n1 = n1 // i  
print(result) 
````

## Ejercicio 14

Implementa un programa que imprima por pantalla todos los **números primos** existentes entre dos números introducidos por teclado, ```n1```y ```n2```, ambos inclusive. El programa deberá comenzar siempre por el menor de ambos, independientemente de en qué orden han sido leídos por teclado.

 

````python
try_again = True; n1 = -1; n2 = -1
while try_again:
  n1 = int(input("Introduce un número positivo mayor a 0: "))
  n2 = int(input("Introduce un número positivo mayor que el anterior: "))
  if(n1 > 0 and n2 > n1): try_again = False 

print("Prime list:")

for i in range(n1, n2+1):
  if i < 2:
    continue

  isPrime = True
  for j in range(2, int(i ** 0.5) + 1):
    if i % j == 0:
      isPrime = False
      break
                
  if isPrime:
    print(i)
````
